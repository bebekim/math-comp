## Step 1 — triangular numbers

`triangular(n)` computes 1 + 2 + ... + n the slow way, one addition at a time.
The proposed formula is n(n+1)/2. Compare them for the first ten values of n.

```python
def triangular(n):
    total = 0
    for k in range(1, n + 1):
        total += k
    return total

predicted = lambda n: n * (n + 1) // 2

if predicted is None:
    print("Fill in `predicted` above, then re-run this cell.")
else:
    for n in range(1, 11):
        observed = triangular(n)
        guess = predicted(n)
        print(n, observed, guess, observed - guess)
```

If you filled in the formula, every gap in the last column is 0. Ten exact matches is
*evidence* for 1 + 2 + ... + n = n(n+1)/2 — but nothing here forces the gap to stay 0
forever. That is what the induction proof on the lesson page is for.

## Step 2 — the sum of cubes

A more dramatic identity:

1^3 + 2^3 + ... + n^3 = (1 + 2 + ... + n)^2 = (n(n+1)/2)^2

Check it exactly, for every n up to 3000. Python integers never overflow, so a
mismatch would be real — and an empty mismatch list is still only a finite observation.

```python
def sum_cubes(n):
    total = 0
    for k in range(1, n + 1):
        total += k**3
    return total

formula = lambda n: (n * (n + 1) // 2) ** 2

if formula is None:
    print("Fill in `formula` above, then re-run this cell.")
else:
    mismatches = []
    for n in range(1, 3001):
        if sum_cubes(n) != formula(n):
            mismatches.append(n)
    print("exact checks:", 3000)
    print("mismatches:", mismatches)
```

## Step 3 — a conjecture that survives 40 trials

The expression n^2 + n + 41 produces a prime for every n from 0 through 39.
Forty successes in a row. Write the expression in and let the loop find where it breaks.

```python
import math

def smallest_divisor(n):
    if n < 2:
        return n
    for d in range(2, math.isqrt(n) + 1):
        if n % d == 0:
            return d
    return None

candidate = lambda n: n*n + n + 41

if candidate is None:
    print("Fill in `candidate` above, then re-run this cell.")
else:
    for n in range(0, 100):
        value = candidate(n)
        divisor = smallest_divisor(value)
        if divisor is not None:
            print("first failure:", n, value, "=", divisor, "x", value // divisor)
            break
```
