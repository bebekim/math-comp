# Breaking chocolate — computer check

An *m* x *n* chocolate bar is broken into its *mn* individual squares, one break at a time (each break turns one piece into two).

You already worked this out on paper: how many breaks are needed, and why.
This notebook checks your reasoning with sympy instead of asking you to trust it.

## Step 1 — describe one break as algebra

Let `p` = number of pieces, `c` = number of breaks so far. One break does:

(p, c) -> (p + 1, c + 1)

```python
from sympy import symbols, simplify, Eq, solve

p, c, m, n = symbols('p c m n')
```

## Step 2 — TRY THIS: show p - c doesn't change

Write the invariant *after* one break, minus the invariant *before* one break.
If your invariant is right, sympy should simplify this to `0`.

```python
change = ((p + 1) - (c + 1)) - (p - c)

print(simplify(change))  # 0 confirms p - c never changes
```

## Step 3 — TRY THIS: solve the general formula

You know `p - c = 1` throughout (check it: 1 piece, 0 breaks at the start).
At the end, `p = m*n`. Use `solve` to find `c` in terms of `m` and `n`.

```python
c_formula = solve(Eq(m * n - c, 1), c)

print(c_formula)  # [m*n - 1]
```
